Задание 1

Переведите содержимое файла purchase_log.txt в словарь purchases вида:
 {'1840e0b9d4': 'Продукты', ...}

Пример работы программы при выводе первых двух элементов словаря purchases:

1840e0b9d4 ‘Продукты‘

4e4f90fcfb  ‘Электроника‘

```python
import json

with open('data/purchase_log.txt', encoding='utf-8') as file:

    file.readline() # пропустить заголовок

    purchases = {}

    for line in file:
        data = json.loads(line)
        key = data['user_id']
        value = data['category']
        purchases[key] = value

    # для вывода данных как в примере работы программы в задаче:
    for user_id, category in list(purchases.items())[:10]:
        print(user_id, f"'{category}'")

```

Задание 2

Для каждого user_id в файле visit_log.csv определите третий столбец с категорией покупки. Если покупка была, сам файл visit_log.csv изменять не надо.

Запишите в файл funnel.csv визиты из файла visit_log.csv, в которых были покупки с указанием категории.

Учтите условия для данных:

содержимое purchase_log.txt помещается в оперативную память компьютера;
содержимое visit_log.csv — нет, используйте только построчную обработку этого файла.

Примеры работы программы при выводе первых трёх строк файла funnel.csv:

user_id,source,category

1840e0b9d4,other,Продукты

4e4f90fcfb,context,Электроника

```python
import csv
import json

with open('data/visit_log.csv', encoding='utf-8') as csv_file, \
     open('data/purchase_log.txt', encoding='utf-8') as txt_file,\
     open('data/funnel.csv', mode='w', encoding='utf-8') as funnel_file:
     
     # пропустить заголовок 
     txt_file.readline() 
     csv_file.readline()

     visit_data = csv.reader(csv_file, delimiter=',')   

     writer = csv.writer(funnel_file)
     writer.writerow(['user_id', 'source', 'category'])

     for user_id, source in visit_data:
          category = purchases.get(user_id)
          if category is not None:
               writer.writerow([user_id, source, category])


```
