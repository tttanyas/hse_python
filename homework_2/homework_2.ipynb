Задание 1

Печатные газеты использовали свой формат дат для каждого выпуска. Для каждой газеты из списка напишите формат указанной даты для перевода в объект datetime:

The Moscow Times - Wednesday, October 2, 2002

The Guardian - Friday, 11.10.13

Daily News - Thursday, 18 August 1977

Пример работы программы

Программа должна выводить на экран объекты типа datetime, соответствующие датам в условии задачи

```python
from datetime import datetime

the_moscow_times_date = 'Wednesday, October 2, 2002'
print(datetime.strptime(the_moscow_times_date, '%A, %B %d, %Y'))

the_guardian_date = 'Friday, 11.10.13'
print(datetime.strptime(the_guardian_date, '%A, %d.%m.%y'))

daily_news = 'Thursday, 18 August 1977'
print(datetime.strptime(daily_news, '%A, %d %B %Y'))
```

Задание 2

Напишите функцию date_range, которая возвращает список дат за период от start_date до end_date. Даты должны вводиться в формате YYYY-MM-DD. В случае неверного формата или при start_date > end_date должен возвращаться пустой список.

Примеры работы программы:

date_range(‘2022-01-01’, ‘2022-01-03’)

[‘2022-01-01’, ‘2022-01-02’, ‘2022-01-03’]

```python
def date_range(start_date, end_date) -> list:

    from datetime import datetime, timedelta

    all_dates = []

    try:

        start_date = datetime.strptime(start_date, '%Y-%m-%d')
        end_date = datetime.strptime(end_date, '%Y-%m-%d')

        if start_date < end_date:
            while start_date <= end_date:
                all_dates.append(start_date.strftime('%Y-%m-%d'))
                start_date += timedelta(days=1)

        else: 
            []

    except:
        []
    
    return all_dates

# проверка
print(date_range('2022-01-01', '2022-01-03'))
print(date_range('2022-01-03', '2022-01-01'))
print(date_range('2022-02-30', '2022-02-31'))

```
